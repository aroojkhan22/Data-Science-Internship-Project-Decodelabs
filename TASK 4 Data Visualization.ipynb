```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")

# cleaning (so you don’t repeat again and again)
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
df = df.drop_duplicates()
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
sns.countplot(x='Survived', data=df)
plt.title("Survival Count")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Count")
plt.show()
```

```python
sns.set_style("whitegrid")
```

```python
sns.countplot(x='Sex', hue='Survived', data=df, palette=['#1f77b4', '#ff4d6d'])
plt.title("Survival by Gender")
plt.xlabel("Sex (0 = Male, 1 = Female)")
plt.ylabel("Count")
plt.show()
```

```python
sns.countplot(x='Pclass', hue='Survived', data=df, palette='viridis')
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.show()
```

```python
sns.histplot(df['Age'], bins=30, kde=True, color='#6a0dad')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(df['Fare'], bins=30, kde=True, color='#2ecc71')
plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Survived', y='Age', data=df, palette=['#ff6b6b', '#4ecdc4'])

plt.title("Age vs Survival")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Age")
plt.show()
```

## 📦 Age vs Survival (Boxplot):

- Shows distribution of age for survived vs not survived passengers
- Helps identify outliers and spread of data
- Some younger passengers had higher survival rates

```python
plt.figure(figsize=(8,5))

sns.boxplot(x='Pclass', y='Fare', data=df, palette='coolwarm')

plt.title("Fare vs Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.show()
```

## 💸 Fare vs Class:

- First-class passengers paid significantly higher fares
- Third-class passengers paid the least
- Clear economic divide is visible
