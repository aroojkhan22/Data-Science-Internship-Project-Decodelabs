# Task 2: Data Cleaning & Preprocessing

## 🎯 Goal:
Prepare the dataset for analysis by cleaning and organizing data.

## 📌 Objectives:
- Handle missing values
- Remove duplicates
- Fix data types
- Convert categorical data into numeric format

---

```python
import pandas as pd

df = pd.read_csv("Titanic-Dataset.csv")  # make sure file is in same folder
```

```python
# Check missing values
df.isnull().sum()
```

## 🔍 Missing Values Analysis:

- Some columns contain missing values.
- "Age" usually has missing values.
- "Embarked" may also have a few missing entries.

```python
# Fix Age
df['Age'] = df['Age'].fillna(df['Age'].mean())

# Fix Embarked
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])
```

```python
df.isnull().sum()
```

```python
# Check duplicates
df.duplicated().sum()
```

```python
# Remove duplicates
df.drop_duplicates(inplace=True)
```

```python
# Convert 'Sex' column
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})
```

## 🧹 Data Cleaning Summary:

- Missing values in "Age" filled with mean
- Missing values in "Embarked" filled with mode
- Duplicate rows removed
- Categorical column "Sex" converted into numeric format

## ✅ Dataset is now clean and ready for analysis
